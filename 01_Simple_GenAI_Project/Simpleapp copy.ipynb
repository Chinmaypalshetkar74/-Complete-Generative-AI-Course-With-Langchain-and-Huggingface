```python
import os
from dotenv import load_dotenv
load_dotenv() ## import environment variables from .env file


os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")

## API KEY FOR Langsmith Tracking
os.environ["LANGSMITH_API_KEY"]= os.getenv("LANGSMITH_API_KEY")

os.environ["LANGCHAIN_TRACING_V2"] = "true"  ## Enable Langchain Tracking V2

os.environ["LANGCHAIN_PROJECT"]= os.getenv("LANGCHAIN_PROJECT")

```

## Data Ingestion

```python
## Data Ingestion--From the website we need to scrape the data
from langchain_community.document_loaders import WebBaseLoader
```

```python
loader=WebBaseLoader("https://ingenero.com/blog/carbon-removal-technologies-for-net-zero/"
                     "https://ingenero.com/blog/ai-in-petroleum-refining-operations/")
loader
```

```python
docs=loader.load()
docs
```

```python
## Loading Data --> Docs --> Divide our text into Chunks (in small pieces of text/doc) --> Convert vectors by Vectors embeddings (Text to vector conversion technique)
## ---> store the vectors in a vector database (We will use FAISS vector database) --> Query the vector database to get the relevant information from the text/documents

from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
docs = text_splitter.split_documents(docs)
```

```python
docs
```

## Embeddings

```python
import os
from dotenv import load_dotenv
load_dotenv() 
```

```python
## Converting all text data into vectors - reason is wheneve work with Q/A chatbot  or rag application we use algorithm like similarity search 
os.environ['HF_TOKEN']=os.getenv("HF_TOKEN")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings=HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

### Vector stores DB - Faiss

```python
from langchain_community.vectorstores import FAISS
vectorstoredb= FAISS.from_documents(docs,embeddings)
```

```python
vectorstoredb
```

```python
## Query From a vector db
query="Every refinery wants to lower its carbon footprint "
result=vectorstoredb.similarity_search(query)
result[0].page_content
```

```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

# Pass model name directly
llm = ChatGoogleGenerativeAI(
    model="gemma-4-31b-it",
    google_api_key=os.getenv("GOOGLE_API_KEY"),
    max_retries=5,
    temperature=0.7
)
```

## Retrieval Chain

```python
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
# crete stuff document chain use for to provide context about doc/document to llm to ans much properly
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    """
Answer the following question based on only provided context:
<context>
{context}
</context>

Question: {input}

"""
)

document_chain = create_stuff_documents_chain(llm,prompt)
document_chain
```

```python
from langchain_core.documents import Document
document_chain.invoke({
    "input":"LangSmith has two usage limits: total traces and extended",
    "context":[Document(page_content= "LangSmith has two usage limits: total traces and extended traces. These correspond to the two metrics we've been tracking on our usage graph. ")]
})
```

However, we want the documents to first come from the retriever we just set up. That way, we can use the retriever to dynamically select the most relevant documents and pass those in for a given question.

```python
## Input --->  Retriver - retriever can be consider as interface (Pathway to get information from vectorstoredb) ---> Vectorstoredb
vectorstoredb
```

```python
retriever = vectorstoredb.as_retriever()
```

```python

from langchain_classic.chains  import create_retrieval_chain

retrieval_chain =create_retrieval_chain(retriever,document_chain)
```

```python
retrieval_chain
```

```python
## Get response from LLm 
response=retrieval_chain.invoke({"input":"LangSmith has two usage limits: total traces and extended"})
response['answer']
```
