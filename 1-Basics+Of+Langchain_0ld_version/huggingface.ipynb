#### Sentence Transformers on Hugging Face
Hugging Face sentence-transformers is a Python framework for state-of-the-art sentence, text and image embeddings. One of the embedding models is used in the HuggingFaceEmbeddings class. We have also added an alias for SentenceTransformerEmbeddings for users who are more familiar with directly using that package.

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
text = "This is a test document. It contains some text that we want to embed using the HuggingFaceEmbeddings class from LangChain."
query_result = embeddings.embed_query(text)
query_result
```

```python
len(query_result)
```

```python
Doc_result = embeddings.embed_documents([text, "This is a test document. It contains some text that we want to embed using the HuggingFaceEmbeddings class from LangChain."])
Doc_result
```
