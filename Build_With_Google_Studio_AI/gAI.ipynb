```python
import google.generativeai as genai
import os
from dotenv import load_dotenv

load_dotenv()

genai.configure(api_key=os.getenv("GOOGLE_API_KEY"))

print("Available models for your API key:")
for model in genai.list_models():
    if "generateContent" in model.supported_generation_methods:
        print(model.name)
```

```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

# Pass model name directly
llm = ChatGoogleGenerativeAI(
    model="gemma-4-31b-it",
    google_api_key=os.getenv("GOOGLE_API_KEY"),
    max_retries=5,
    temperature=0.7
)

response = llm.invoke("Hello, working test!")
print(response.content)
```

```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

# Pass model name directly
llm = ChatGoogleGenerativeAI(
    model="gemini-flash-lite-latest",
    google_api_key=os.getenv("GOOGLE_API_KEY"),
    max_retries=5,
    temperature=0.7
)

response = llm.invoke("2028-29 top upcoming high paying careers in the world not affected by AI and automation, with a brief description of each career.")
print(response.content)
```

```python
## Chat Prompt template for Google Gemini LLM

from langchain_core.prompts import ChatPromptTemplate


prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful AI assistant to guide users."),
        ("user", "{input}"),
    ]
)
prompt
```

```python
chain = prompt | llm

response = chain.invoke({"input": "What is langsmith?"})

print(response)
```

```python
### StrongOutput Parser for Google Gemini LLM - use of Langchain Output Parser to parse the output of Google Gemini LLM
from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
chain = prompt | llm | output_parser

response = chain.invoke({"input": "What is langsmith?"})
print(response)
```
